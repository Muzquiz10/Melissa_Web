# Optimizar catálogo PDF para móvil

1. Cambia `PDF_ORIGINAL` por la ruta del catálogo que quieras optimizar.
2. Si quieres elegir el nombre de salida, escribe una ruta en `PDF_SALIDA`; si lo dejas en `None`, se creará automáticamente con el sufijo `_mobile.pdf`.
3. Ejecuta las celdas en orden.

```python
PDF_ORIGINAL = r"assets/Catalogo_2026-2027.pdf"
PDF_SALIDA = None

DPI = 144
QUALITY = 78
AUTO_INSTALL = True
```

```python
from pathlib import Path

from optimizar_catalogo_movil import default_output_path, format_mb, optimize_pdf

input_path = Path(PDF_ORIGINAL).expanduser().resolve()
output_path = Path(PDF_SALIDA).expanduser().resolve() if PDF_SALIDA else default_output_path(input_path)

if not input_path.exists():
    raise FileNotFoundError(f"No existe el archivo: {input_path}")

print(f"PDF original:   {input_path}")
print(f"PDF optimizado: {output_path}")
print(f"DPI: {DPI} | Calidad: {QUALITY}")

optimize_pdf(
    input_path=input_path,
    output_path=output_path,
    dpi=DPI,
    quality=QUALITY,
    auto_install=AUTO_INSTALL,
)

original_size = input_path.stat().st_size
optimized_size = output_path.stat().st_size

print("\nListo.")
print(f"Original:   {format_mb(original_size)}")
print(f"Optimizado: {format_mb(optimized_size)}")
print(f"Ahorro:     {format_mb(original_size - optimized_size)}")
```
